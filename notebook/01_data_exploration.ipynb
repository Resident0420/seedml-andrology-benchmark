```python
# ============================================
# 1. Import Libraries
# ============================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================
# 2. Load Dataset
# ============================================
df = pd.read_csv(
    "../data/SEED-ML/infertility_man_data-v2.csv",
    sep=";"
)

# Preview the dataset
df.head()
```

```python
# ============================================
# 3. Inspect the Target Variable
# ============================================

# Class distribution (counts)
print(df["diagnostic"].value_counts())

# Class distribution (percentages)
print((df["diagnostic"].value_counts(normalize=True) * 100).round(2))

# Verify available diagnostic classes
print(df["diagnostic"].unique())

# Preview the first five records
df.head().T
```

```python
# ============================================
# 4. Dataset Summary
# ============================================

summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isna().sum(),
    "Unique Values": df.nunique()
})

summary

# ============================================
# Exploratory Data Inspection (Optional)
# ============================================

for column in df.columns:
    print("=" * 80)
    print(column)
    print(df[column].value_counts().head(10))
    print()
```
