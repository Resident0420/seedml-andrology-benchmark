```python
# ============================================
# 1. Import Libraries
# ============================================
import sys
from pathlib import Path

import numpy as np
import pandas as pd

# ============================================
# 2. Configure Project Path
# ============================================
PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

# ============================================
# 3. Import Project Modules
# ============================================
from src.config import TARGET, TIER1_FEATURES
from src.preprocessing import (
    replace_false_with_nan,
    summarize_missingness,
    identify_numeric_columns,
    identify_categorical_columns,
    convert_numeric_columns,
    dataset_summary,
)

# ============================================
# 4. Load Dataset
# ============================================
df = pd.read_csv(
    PROJECT_ROOT / "data" / "SEED-ML" / "infertility_man_data-v2.csv",
    sep=";",
)
```

```python
# ============================================
# 5. Extract Tier 1 Features
# ============================================

X = df[TIER1_FEATURES].copy()
y = df[TARGET].copy()

# ============================================
# 6. Identify "False" Encoded Missing Values
# ============================================

false_summary = pd.DataFrame({
    "False Count": (X == "False").sum(),
    "False (%)": ((X == "False").mean() * 100).round(2),
}).sort_values("False (%)", ascending=False)

false_summary
```

```python
# ============================================
# 7. Clean Tier 1 Features
# ============================================

# Replace encoded "False" values with missing values
X_clean = replace_false_with_nan(X)

# Summarize missingness after cleaning
missing_summary = summarize_missingness(X_clean)
missing_summary

# ============================================
# 8. Identify Variable Types
# ============================================

numeric_columns = identify_numeric_columns(X_clean)

categorical_columns = identify_categorical_columns(
    X_clean,
    numeric_columns,
)

print("Numeric columns:", numeric_columns)
print("Categorical columns:", categorical_columns)

# ============================================
# 9. Convert Numeric Features
# ============================================

X_clean = convert_numeric_columns(
    X_clean,
    numeric_columns,
)

# Inspect cleaned dataset
dataset_summary(X_clean)

# Verify final data types
X_clean.dtypes
```
